### 3.3.1 Eğitilebilir ağırlıkları olmayan basit bir öz-dikkat mekanizması

- Bu bölüm, hiçbir eğitilebilir ağırlık içermeyen, çok basitleştirilmiş bir öz-dikkat varyantını açıklar
- Bu tamamen gösterim amaçlıdır ve transformer'larda kullanılan dikkat mekanizması DEĞİLDİR
- Bir sonraki bölüm, 3.3.2, bu basit dikkat mekanizmasını gerçek öz-dikkat mekanizmasını uygulayacak şekilde genişletecek
- Elimizde $x^{(1)}$ ile $x^{(T)}$ arasında bir girdi dizisi olduğunu varsayalım
  - Girdi, 2. bölümde anlatıldığı gibi zaten token gömmelerine dönüştürülmüş bir metindir (örneğin "Your journey starts with one step" gibi bir cümle)
  - Örneğin $x^{(1)}$, "Your" kelimesini temsil eden d boyutlu bir vektördür ve bu böyle devam eder
- **Amaç:** $x^{(1)}$ ile $x^{(T)}$ arasındaki her $x^{(i)}$ girdi dizisi elemanı için bağlam vektörleri $z^{(i)}$ hesaplamak ($z$ ve $x$ aynı boyuttadır)
    - Bir bağlam vektörü $z^{(i)}$, $x^{(1)}$ ile $x^{(T)}$ girdileri üzerinden ağırlıklı bir toplamdır
    - Bağlam vektörü, belirli bir girdiye özgü olarak "bağlamsaldır"
      - Rastgele bir girdi token'ı için yer tutucu olan $x^{(i)}$ yerine, ikinci girdiyi, $x^{(2)}$ değerini ele alalım
      - Somut bir örnekle devam etmek için, yer tutucu $z^{(i)}$ yerine ikinci çıktı bağlam vektörünü, $z^{(2)}$ değerini ele alıyoruz
      - İkinci bağlam vektörü $z^{(2)}$, ikinci girdi elemanı $x^{(2)}$ dikkate alınarak ağırlıklandırılmış, $x^{(1)}$ ile $x^{(T)}$ arasındaki tüm girdiler üzerinden ağırlıklı bir toplamdır
      - Dikkat ağırlıkları, $z^{(2)}$ hesaplanırken girdi elemanlarının her birinin ağırlıklı toplama ne kadar katkıda bulunacağını belirleyen ağırlıklardır
      - Kısacası $z^{(2)}$ değerini, eldeki göreve ilgili diğer tüm girdi elemanlarına ait bilgiyi de içeren, $x^{(2)}$ değerinin değiştirilmiş bir sürümü olarak düşünün

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/07.webp" width="400px">

- (Görsel karmaşayı azaltmak için bu şekildeki sayıların virgülden sonra tek haneye kırpıldığını unutmayın; benzer şekilde diğer şekiller de kırpılmış değerler içerebilir)

- Yerleşik kullanımda, normalleştirilmemiş dikkat ağırlıklarına **"dikkat skorları"** (attention scores), toplamı 1 olan normalleştirilmiş dikkat skorlarına ise **"dikkat ağırlıkları"** (attention weights) denir

- Aşağıdaki kod, yukarıdaki şekli adım adım izler

<br>

- **1. Adım:** normalleştirilmemiş dikkat skorlarını $\omega$ hesapla
- İkinci girdi token'ını sorgu olarak kullandığımızı varsayalım, yani $q^{(2)} = x^{(2)}$; normalleştirilmemiş dikkat skorlarını nokta çarpımlarıyla hesaplarız:
    - $\omega_{21} = x^{(1)} q^{(2)\top}$
    - $\omega_{22} = x^{(2)} q^{(2)\top}$
    - $\omega_{23} = x^{(3)} q^{(2)\top}$
    - ...
    - $\omega_{2T} = x^{(T)} q^{(2)\top}$
- Yukarıda $\omega$, normalleştirilmemiş dikkat skorlarını simgelemek için kullanılan Yunan harfi "omega"dır
    - $\omega_{21}$ içindeki "21" alt simgesi, 2 numaralı girdi dizisi elemanının 1 numaralı girdi dizisi elemanına karşı sorgu olarak kullanıldığı anlamına gelir

- 3. bölümde anlatıldığı gibi 3 boyutlu vektörlere zaten gömülmüş şu girdi cümlesine sahip olduğumuzu varsayalım (gösterim amacıyla burada çok küçük bir gömme boyutu kullanıyoruz, böylece satır sonu olmadan sayfaya sığıyor):

```python
import torch

inputs = torch.tensor(
  [[0.43, 0.15, 0.89], # Your     (x^1)
   [0.55, 0.87, 0.66], # journey  (x^2)
   [0.57, 0.85, 0.64], # starts   (x^3)
   [0.22, 0.58, 0.33], # with     (x^4)
   [0.77, 0.25, 0.10], # one      (x^5)
   [0.05, 0.80, 0.55]] # step     (x^6)
)
```

- (Bu kitapta, eğitim örneklerinin satır, öznitelik değerlerinin sütun olarak temsil edildiği yaygın makine öğrenmesi ve derin öğrenme kuralını izliyoruz; yukarıda gösterilen tensörde her satır bir kelimeyi, her sütun bir gömme boyutunu temsil eder)

- Bu bölümün başlıca amacı, ikinci girdi dizisi $x^{(2)}$ sorgu olarak kullanılarak bağlam vektörü $z^{(2)}$ değerinin nasıl hesaplandığını göstermektir

- Şekil, bu sürecin ilk adımını gösterir; bu adım, $x^{(2)}$ ile diğer tüm girdi elemanları arasındaki dikkat skorlarının ω bir nokta çarpımı işlemiyle hesaplanmasını içerir

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/08.webp" width="400px">

- Bağlam vektörü $z^{(2)}$ değerini hesaplamak için örnek olarak 2 numaralı girdi dizisi elemanını, $x^{(2)}$ değerini kullanıyoruz; bu bölümün ilerleyen kısımlarında bunu tüm bağlam vektörlerini hesaplayacak şekilde genelleştireceğiz.
- İlk adım, sorgu $x^{(2)}$ ile diğer tüm girdi token'ları arasındaki nokta çarpımını hesaplayarak normalleştirilmemiş dikkat skorlarını bulmaktır:

```python
query = inputs[1]  # 2nd input token is the query

attn_scores_2 = torch.empty(inputs.shape[0])
for i, x_i in enumerate(inputs):
    attn_scores_2[i] = torch.dot(x_i, query) # dot product (transpose not necessary here since they are 1-dim vectors)

print(attn_scores_2)
```

- Yan not: nokta çarpımı esasen iki vektörü eleman bazında çarpıp elde edilen çarpımları toplamanın kısa yoludur:

```python
res = 0.

for idx, element in enumerate(inputs[0]):
    res += inputs[0][idx] * query[idx]

print(res)
print(torch.dot(inputs[0], query))
```

- **2. Adım:** normalleştirilmemiş dikkat skorlarını ("omega"lar, $\omega$) toplamları 1 olacak şekilde normalleştir
- Normalleştirilmemiş dikkat skorlarının toplamını 1 yapmanın basit bir yolu (yorumlama açısından faydalı ve eğitim kararlılığı için önemli olan yerleşik bir uygulama):

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/09.webp" width="500px">

```python
attn_weights_2_tmp = attn_scores_2 / attn_scores_2.sum()

print("Attention weights:", attn_weights_2_tmp)
print("Sum:", attn_weights_2_tmp.sum())
```

- Ancak pratikte, uç değerleri daha iyi ele alan ve eğitim sırasında daha elverişli gradyan özelliklerine sahip olan softmax fonksiyonunu normalleştirme için kullanmak yaygındır ve önerilir.
- Aşağıda, ölçekleme için softmax fonksiyonunun yalın bir uygulaması yer alıyor; bu da vektör elemanlarını toplamları 1 olacak şekilde normalleştirir:

```python
def softmax_naive(x):
    return torch.exp(x) / torch.exp(x).sum(dim=0)

attn_weights_2_naive = softmax_naive(attn_scores_2)

print("Attention weights:", attn_weights_2_naive)
print("Sum:", attn_weights_2_naive.sum())
```

- Yukarıdaki yalın uygulama, taşma (overflow) ve alt taşma (underflow) sorunları nedeniyle çok büyük veya çok küçük girdi değerlerinde sayısal kararsızlık sorunları yaşayabilir
- Bu nedenle pratikte, başarım için ileri düzeyde optimize edilmiş olan PyTorch'un softmax uygulamasını kullanmak önerilir:

```python
attn_weights_2 = torch.softmax(attn_scores_2, dim=0)

print("Attention weights:", attn_weights_2)
print("Sum:", attn_weights_2.sum())
```

- **3. Adım**: gömülmüş girdi token'larını $x^{(i)}$ dikkat ağırlıklarıyla çarpıp elde edilen vektörleri toplayarak bağlam vektörü $z^{(2)}$ değerini hesapla:

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/10.webp" width="500px">

```python
query = inputs[1] # 2nd input token is the query

context_vec_2 = torch.zeros(query.shape)
for i,x_i in enumerate(inputs):
    context_vec_2 += attn_weights_2[i]*x_i

print(context_vec_2)
```

### 3.3.2 Tüm girdi token'ları için dikkat ağırlıklarını hesaplamak

#### Tüm girdi dizisi token'larına genelleştirme:

- Yukarıda, 2 numaralı girdi için dikkat ağırlıklarını ve bağlam vektörünü hesapladık (aşağıdaki şekilde vurgulanan satırda gösterildiği gibi)
- Şimdi bu hesaplamayı, tüm dikkat ağırlıklarını ve bağlam vektörlerini hesaplayacak şekilde genelleştiriyoruz

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/11.webp" width="400px">

- (Görsel karmaşayı azaltmak için bu şekildeki sayıların virgülden sonra iki haneye kırpıldığını unutmayın; her satırdaki değerlerin toplamı 1,0 yani %100 olmalıdır; diğer şekillerdeki haneler de benzer şekilde kırpılmıştır)

- Öz-dikkatte süreç, dikkat skorlarının hesaplanmasıyla başlar; bunlar daha sonra toplamı 1 olan dikkat ağırlıklarını elde etmek için normalleştirilir
- Ardından bu dikkat ağırlıkları, girdilerin ağırlıklı toplamı yoluyla bağlam vektörlerini üretmek için kullanılır

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/12.webp" width="400px">

- Normalleştirilmemiş dikkat skoru matrisini hesaplamak için önceki **1. adımı** tüm ikili elemanlara uygula:

```python
attn_scores = torch.empty(6, 6)

for i, x_i in enumerate(inputs):
    for j, x_j in enumerate(inputs):
        attn_scores[i, j] = torch.dot(x_i, x_j)

print(attn_scores)
```

- Yukarıdakinin aynısını matris çarpımı yoluyla daha verimli biçimde elde edebiliriz:

```python
attn_scores = inputs @ inputs.T
print(attn_scores)
```

- Daha önceki **2. adıma** benzer şekilde, her satırdaki değerlerin toplamı 1 olacak biçimde her satırı normalleştiriyoruz:

```python
attn_weights = torch.softmax(attn_scores, dim=-1)
print(attn_weights)
```

- Her satırdaki değerlerin gerçekten 1'e toplandığının hızlı bir doğrulaması:

```python
row_2_sum = sum([0.1385, 0.2379, 0.2333, 0.1240, 0.1082, 0.1581])
print("Row 2 sum:", row_2_sum)

print("All row sums:", attn_weights.sum(dim=-1))
```

- Tüm bağlam vektörlerini hesaplamak için önceki **3. adımı** uygula:

```python
all_context_vecs = attn_weights @ inputs
print(all_context_vecs)
```

- Bir sağlama olarak, daha önce hesaplanan bağlam vektörü $z^{(2)} = [0.4419, 0.6515, 0.5683]$ değerini yukarıdaki 2. satırda bulabiliriz:

```python
print("Previous 2nd context vector:", context_vec_2)
```

### 3.4.1 Dikkat ağırlıklarını adım adım hesaplamak

- Bu bölümde, orijinal transformer mimarisinde, GPT modellerinde ve popüler diğer LLM'lerin çoğunda kullanılan öz-dikkat mekanizmasını uyguluyoruz
- Bu öz-dikkat mekanizmasına "ölçeklenmiş nokta çarpımı dikkati" (scaled dot-product attention) de denir
- Genel fikir öncekiyle benzerdir:
  - Belirli bir girdi elemanına özgü olarak, girdi vektörleri üzerinden ağırlıklı toplamlar biçiminde bağlam vektörleri hesaplamak istiyoruz
  - Bunun için dikkat ağırlıklarına ihtiyacımız var
- Göreceğiniz gibi, daha önce tanıtılan temel dikkat mekanizmasına kıyasla yalnızca küçük farklar var:
  - En dikkat çekici fark, model eğitimi sırasında güncellenen ağırlık matrislerinin devreye girmesidir
  - Bu eğitilebilir ağırlık matrisleri, modelin (özellikle model içindeki dikkat modülünün) "iyi" bağlam vektörleri üretmeyi öğrenebilmesi için kritiktir

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/14.webp" width="600px">

- Öz-dikkat mekanizmasını adım adım uygularken, önce üç eğitilebilir ağırlık matrisini $W_q$, $W_k$ ve $W_v$ tanıtacağız
- Bu üç matris, gömülmüş girdi token'larını $x^{(i)}$ matris çarpımı yoluyla sorgu, anahtar ve değer vektörlerine izdüşürmek için kullanılır:

  - Sorgu vektörü: $q^{(i)} = x^{(i)}\,W_q $
  - Anahtar vektörü: $k^{(i)} = x^{(i)}\,W_k $
  - Değer vektörü: $v^{(i)} = x^{(i)}\,W_v $



- Girdi $x$ ile sorgu vektörü $q$ değerlerinin gömme boyutları, modelin tasarımına ve belirli uygulamaya bağlı olarak aynı ya da farklı olabilir
- GPT modellerinde girdi ve çıktı boyutları genellikle aynıdır; ancak gösterim amacıyla, hesaplamayı daha iyi izleyebilmek için burada farklı girdi ve çıktı boyutları seçiyoruz:

```python
x_2 = inputs[1] # second input element
d_in = inputs.shape[1] # the input embedding size, d=3
d_out = 2 # the output embedding size, d=2
```

- Aşağıda üç ağırlık matrisini başlatıyoruz; gösterim amacıyla çıktılardaki karmaşayı azaltmak için `requires_grad=False` ayarladığımızı unutmayın; ancak ağırlık matrislerini model eğitimi için kullanacak olsaydık, bu matrisleri eğitim sırasında güncellemek için `requires_grad=True` ayarlardık

```python
torch.manual_seed(123)

W_query = torch.nn.Parameter(torch.rand(d_in, d_out), requires_grad=False)
W_key   = torch.nn.Parameter(torch.rand(d_in, d_out), requires_grad=False)
W_value = torch.nn.Parameter(torch.rand(d_in, d_out), requires_grad=False)
```

- Ardından sorgu, anahtar ve değer vektörlerini hesaplıyoruz:

```python
query_2 = x_2 @ W_query # _2 because it's with respect to the 2nd input element
key_2 = x_2 @ W_key 
value_2 = x_2 @ W_value

print(query_2)
```

- Aşağıda görebileceğimiz gibi, 6 girdi token'ını 3 boyutlu uzaydan 2 boyutlu bir gömme uzayına başarıyla izdüşürdük:

```python
keys = inputs @ W_key 
values = inputs @ W_value

print("keys.shape:", keys.shape)
print("values.shape:", values.shape)
```

- Sonraki adımda, **2. adımda**, sorgu ile her anahtar vektörü arasındaki nokta çarpımını hesaplayarak normalleştirilmemiş dikkat skorlarını buluyoruz:

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/15.webp" width="600px">

```python
keys_2 = keys[1] # Python starts index at 0
attn_score_22 = query_2.dot(keys_2)
print(attn_score_22)
```

- 6 girdimiz olduğu için, verilen sorgu vektörü için 6 dikkat skorumuz var:

```python
attn_scores_2 = query_2 @ keys.T # All attention scores for given query
print(attn_scores_2)
```

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/16.webp" width="600px">

- Ardından, **3. adımda**, daha önce kullandığımız softmax fonksiyonuyla dikkat ağırlıklarını (toplamı 1 olan normalleştirilmiş dikkat skorları) hesaplıyoruz
- Öncekinden farkı, artık dikkat skorlarını gömme boyutunun kareköküne $\sqrt{d_k}$ (yani `d_k**0.5`) bölerek ölçeklendirmemizdir:

```python
d_k = keys.shape[1]
attn_weights_2 = torch.softmax(attn_scores_2 / d_k**0.5, dim=-1)
print(attn_weights_2)
```

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/17.webp" width="600px">

- **4. adımda** artık 2 numaralı girdi sorgu vektörü için bağlam vektörünü hesaplıyoruz:

```python
context_vec_2 = attn_weights_2 @ values
print(context_vec_2)
```

### 3.4.2 Derli toplu bir SelfAttention sınıfı uygulamak

- Hepsini bir araya getirerek öz-dikkat mekanizmasını şu şekilde uygulayabiliriz:

```python
import torch.nn as nn

class SelfAttention_v1(nn.Module):

    def __init__(self, d_in, d_out):
        super().__init__()
        self.W_query = nn.Parameter(torch.rand(d_in, d_out))
        self.W_key   = nn.Parameter(torch.rand(d_in, d_out))
        self.W_value = nn.Parameter(torch.rand(d_in, d_out))

    def forward(self, x):
        keys = x @ self.W_key
        queries = x @ self.W_query
        values = x @ self.W_value
        
        attn_scores = queries @ keys.T # omega
        attn_weights = torch.softmax(
            attn_scores / keys.shape[-1]**0.5, dim=-1
        )

        context_vec = attn_weights @ values
        return context_vec

torch.manual_seed(123)
sa_v1 = SelfAttention_v1(d_in, d_out)
print(sa_v1(inputs))
```

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/18.webp" width="400px">

- Yukarıdaki uygulamayı, bias birimlerini devre dışı bıraktığımızda matris çarpımına eşdeğer olan PyTorch'un Linear katmanlarını kullanarak sadeleştirebiliriz
- `nn.Linear` kullanmanın elle yazdığımız `nn.Parameter(torch.rand(...)` yaklaşımına göre bir diğer büyük avantajı, `nn.Linear` sınıfının tercih edilen bir ağırlık başlatma şemasına sahip olması ve bunun daha kararlı model eğitimine yol açmasıdır

```python
class SelfAttention_v2(nn.Module):

    def __init__(self, d_in, d_out, qkv_bias=False):
        super().__init__()
        self.W_query = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.W_key   = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.W_value = nn.Linear(d_in, d_out, bias=qkv_bias)

    def forward(self, x):
        keys = self.W_key(x)
        queries = self.W_query(x)
        values = self.W_value(x)
        
        attn_scores = queries @ keys.T
        attn_weights = torch.softmax(attn_scores / keys.shape[-1]**0.5, dim=-1)

        context_vec = attn_weights @ values
        return context_vec

torch.manual_seed(789)
sa_v2 = SelfAttention_v2(d_in, d_out)
print(sa_v2(inputs))
```

- `SelfAttention_v1` ve `SelfAttention_v2` sınıflarının farklı çıktılar vermesinin nedeni, ağırlık matrisleri için farklı başlangıç ağırlıkları kullanmalarıdır

## 3.5 Nedensel dikkatle gelecekteki kelimeleri gizlemek

### 3.5.1 Nedensel dikkat maskesi uygulamak

- Bu bölümde, önceki öz-dikkat mekanizmasını nedensel bir öz-dikkat mekanizmasına dönüştürüyoruz
- Nedensel öz-dikkat, modelin bir dizideki belirli bir konum için yaptığı tahminin yalnızca önceki konumlardaki bilinen çıktılara bağlı olmasını, gelecekteki konumlara bağlı olmamasını sağlar
- Daha basit ifadeyle, bu her bir sonraki kelime tahmininin yalnızca önceki kelimelere bağlı olmasını güvence altına alır
- Bunu başarmak için, verilen her token'da gelecekteki token'ları (girdi metninde mevcut token'dan sonra gelenleri) maskeliyoruz:

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/20.webp" width="600px">

- Nedensel öz-dikkati göstermek ve uygulamak için, önceki bölümdeki dikkat skorları ve ağırlıklarıyla çalışalım:

```python
# Kolaylık olsun diye önceki bölümdeki SelfAttention_v2
# nesnesinin sorgu ve anahtar ağırlık matrislerini yeniden kullanıyoruz
queries = sa_v2.W_query(inputs)
keys = sa_v2.W_key(inputs) 
attn_scores = queries @ keys.T

attn_weights = torch.softmax(attn_scores / keys.shape[-1]**0.5, dim=-1)
print(attn_weights)
```

- Gelecekteki dikkat ağırlıklarını maskelemenin en basit yolu, PyTorch'un tril fonksiyonuyla ana köşegenin altındaki (köşegen dahil) elemanları 1, üstündekileri 0 olan bir maske oluşturmaktır:

```python
context_length = attn_scores.shape[0]
mask_simple = torch.tril(torch.ones(context_length, context_length))
print(mask_simple)
```

- Ardından, köşegenin üzerindeki dikkat skorlarını sıfırlamak için dikkat ağırlıklarını bu maskeyle çarpabiliriz:

```python
masked_simple = attn_weights*mask_simple
print(masked_simple)
```

- Ancak maske yukarıdaki gibi softmax'tan sonra uygulanırsa, softmax'ın oluşturduğu olasılık dağılımını bozar
- Softmax, tüm çıktı değerlerinin toplamının 1 olmasını sağlar
- Softmax'tan sonra maskeleme, çıktıların toplamının yeniden 1 olması için tekrar normalleştirilmesini gerektirir; bu da süreci karmaşıklaştırır ve istenmeyen etkilere yol açabilir

- Satırların toplamının 1 olduğundan emin olmak için dikkat ağırlıklarını şu şekilde normalleştirebiliriz:

```python
row_sums = masked_simple.sum(dim=-1, keepdim=True)
masked_simple_norm = masked_simple / row_sums
print(masked_simple_norm)
```

- Teknik olarak nedensel dikkat mekanizmasını kodlamayı bitirmiş olsak da, aynı sonucu elde etmenin daha verimli bir yoluna kısaca bakalım
- Yani köşegenin üzerindeki dikkat ağırlıklarını sıfırlayıp sonuçları yeniden normalleştirmek yerine, köşegenin üzerindeki normalleştirilmemiş dikkat skorlarını softmax fonksiyonuna girmeden önce eksi sonsuzla maskeleyebiliriz:

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/21.webp" width="450px">

```python
mask = torch.triu(torch.ones(context_length, context_length), diagonal=1)
masked = attn_scores.masked_fill(mask.bool(), -torch.inf)
print(masked)
```

- Aşağıda görebileceğimiz gibi, artık her satırdaki dikkat ağırlıkları yeniden doğru şekilde 1'e toplanıyor:

```python
attn_weights = torch.softmax(masked / keys.shape[-1]**0.5, dim=-1)
print(attn_weights)
```

### 3.5.2 Dropout ile ek dikkat ağırlıklarını maskelemek

- Buna ek olarak, eğitim sırasında aşırı öğrenmeyi azaltmak için dropout da uyguluyoruz
- Dropout birkaç yerde uygulanabilir:
  - örneğin dikkat ağırlıkları hesaplandıktan sonra;
  - ya da dikkat ağırlıkları değer vektörleriyle çarpıldıktan sonra
- Burada dropout maskesini dikkat ağırlıkları hesaplandıktan sonra uygulayacağız, çünkü bu daha yaygındır

- Ayrıca bu özel örnekte %50'lik bir dropout oranı kullanıyoruz; bu, dikkat ağırlıklarının yarısının rastgele maskelenmesi anlamına gelir. (İlerideki bölümlerde GPT modelini eğitirken 0,1 veya 0,2 gibi daha düşük bir dropout oranı kullanacağız.)

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/22.webp" width="400px">

- 0,5 (%50) dropout oranı uygularsak, düşürülmeyen değerler buna göre 1/0,5 = 2 çarpanıyla ölçeklenir
- Ölçekleme, 1 / (1 - `dropout_rate`) formülüyle hesaplanır

```python
torch.manual_seed(123)
dropout = torch.nn.Dropout(0.5) # dropout rate of 50%
example = torch.ones(6, 6) # create a matrix of ones

print(dropout(example))
```

```python
torch.manual_seed(123)
print(dropout(attn_weights))
```

### 3.5.3 Derli toplu bir nedensel öz-dikkat sınıfı uygulamak

- Artık nedensel maske ve dropout maskesi dahil, çalışan bir öz-dikkat uygulaması yazmaya hazırız
- Bir şey daha var: `CausalAttention` sınıfımızın 2. bölümde uyguladığımız veri yükleyicinin ürettiği yığın çıktılarını desteklemesi için, birden fazla girdiden oluşan yığınları işleyecek kodu da yazmalıyız
- Basitlik adına, böyle bir yığın girdisini benzetmek için girdi metni örneğini çoğaltıyoruz:

```python
batch = torch.stack((inputs, inputs), dim=0)
print(batch.shape) # 2 inputs with 6 tokens each, and each token has embedding dimension 3
```

```python
class CausalAttention(nn.Module):

    def __init__(self, d_in, d_out, context_length,
                 dropout, qkv_bias=False):
        super().__init__()
        self.d_out = d_out
        self.W_query = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.W_key   = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.W_value = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.dropout = nn.Dropout(dropout) # New
        self.register_buffer('mask', torch.triu(torch.ones(context_length, context_length), diagonal=1)) # New

    def forward(self, x):
        b, num_tokens, d_in = x.shape # New batch dimension b
        # `num_tokens` değerinin `context_length` değerini aştığı girdilerde, bu durum
        # aşağıdaki maske oluşturmada hatalara yol açar.
        # Pratikte bu bir sorun değildir; çünkü LLM (4-7. bölümler) girdilerin bu forward
        # metoduna ulaşmadan önce `context_length` değerini aşmamasını sağlar.
        keys = self.W_key(x)
        queries = self.W_query(x)
        values = self.W_value(x)

        attn_scores = queries @ keys.transpose(1, 2) # Changed transpose
        attn_scores.masked_fill_(  # New, _ ops are in-place
            self.mask.bool()[:num_tokens, :num_tokens], -torch.inf)  # `:num_tokens` to account for cases where the number of tokens in the batch is smaller than the supported context_size
        attn_weights = torch.softmax(
            attn_scores / keys.shape[-1]**0.5, dim=-1
        )
        attn_weights = self.dropout(attn_weights) # New

        context_vec = attn_weights @ values
        return context_vec

torch.manual_seed(123)

context_length = batch.shape[1]
ca = CausalAttention(d_in, d_out, context_length, 0.0)

context_vecs = ca(batch)

print(context_vecs)
print("context_vecs.shape:", context_vecs.shape)
```

- Dropout'un yalnızca eğitim sırasında uygulandığını, çıkarım sırasında uygulanmadığını unutmayın

### 3.6.1 Birden çok tek başlı dikkat katmanını üst üste yığmak

- Aşağıda, daha önce uygulanan öz-dikkatin bir özeti yer alıyor (basitlik adına nedensel ve dropout maskeleri gösterilmemiştir)

- Buna tek başlı dikkat (single-head attention) da denir:

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/24.webp" width="400px">

- Bir çok başlı dikkat modülü elde etmek için birden çok tek başlı dikkat modülünü basitçe üst üste yığarız:

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/25.webp" width="400px">

- Çok başlı dikkatin arkasındaki temel fikir, dikkat mekanizmasını farklı, öğrenilmiş doğrusal izdüşümlerle birden çok kez (paralel olarak) çalıştırmaktır. Bu, modelin farklı konumlardaki farklı temsil alt uzaylarından gelen bilgilere birlikte dikkat etmesini sağlar.

```python
class MultiHeadAttentionWrapper(nn.Module):

    def __init__(self, d_in, d_out, context_length, dropout, num_heads, qkv_bias=False):
        super().__init__()
        self.heads = nn.ModuleList(
            [CausalAttention(d_in, d_out, context_length, dropout, qkv_bias) 
             for _ in range(num_heads)]
        )

    def forward(self, x):
        return torch.cat([head(x) for head in self.heads], dim=-1)


torch.manual_seed(123)

context_length = batch.shape[1] # This is the number of tokens
d_in, d_out = 3, 2
mha = MultiHeadAttentionWrapper(
    d_in, d_out, context_length, 0.0, num_heads=2
)

context_vecs = mha(batch)

print(context_vecs)
print("context_vecs.shape:", context_vecs.shape)
```

- Yukarıdaki uygulamada gömme boyutu 4'tür; çünkü anahtar, sorgu ve değer vektörleri ile bağlam vektörü için gömme boyutu olarak `d_out=2` kullandık. Ve 2 dikkat başımız olduğu için çıktı gömme boyutu 2*2=4 olur

&nbsp;
### 3.6.2 Ağırlık bölmeleriyle çok başlı dikkat uygulamak

- Yukarıdaki, çok başlı dikkatin sezgisel ve tamamen işlevsel bir uygulaması olsa da (daha önceki tek başlı `CausalAttention` uygulamasını saran bir yapı), aynı sonucu elde etmek için `MultiHeadAttention` adında bağımsız bir sınıf yazabiliriz

- Bu bağımsız `MultiHeadAttention` sınıfında tek tek dikkat başlarını birleştirmiyoruz
- Bunun yerine tek bir W_query, W_key ve W_value ağırlık matrisi oluşturup, bunları her dikkat başı için ayrı matrislere bölüyoruz:

```python
class MultiHeadAttention(nn.Module):
    def __init__(self, d_in, d_out, context_length, dropout, num_heads, qkv_bias=False):
        super().__init__()
        assert (d_out % num_heads == 0), \
            "d_out must be divisible by num_heads"

        self.d_out = d_out
        self.num_heads = num_heads
        self.head_dim = d_out // num_heads # Reduce the projection dim to match desired output dim

        self.W_query = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.W_key = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.W_value = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.out_proj = nn.Linear(d_out, d_out)  # Linear layer to combine head outputs
        self.dropout = nn.Dropout(dropout)
        self.register_buffer(
            "mask",
            torch.triu(torch.ones(context_length, context_length),
                       diagonal=1)
        )

    def forward(self, x):
        b, num_tokens, d_in = x.shape
        # `CausalAttention` sınıfında olduğu gibi, `num_tokens` değerinin `context_length`
        # değerini aştığı girdilerde bu durum aşağıdaki maske oluşturmada hatalara yol açar.
        # Pratikte bu bir sorun değildir; çünkü LLM (4-7. bölümler) girdilerin bu forward
        # metoduna ulaşmadan önce `context_length` değerini aşmamasını sağlar.

        keys = self.W_key(x) # Shape: (b, num_tokens, d_out)
        queries = self.W_query(x)
        values = self.W_value(x)

        # Matrisi, bir `num_heads` boyutu ekleyerek örtük olarak bölüyoruz
        # Son boyutu aç: (b, num_tokens, d_out) -> (b, num_tokens, num_heads, head_dim)
        keys = keys.view(b, num_tokens, self.num_heads, self.head_dim) 
        values = values.view(b, num_tokens, self.num_heads, self.head_dim)
        queries = queries.view(b, num_tokens, self.num_heads, self.head_dim)

        # Devrik al: (b, num_tokens, num_heads, head_dim) -> (b, num_heads, num_tokens, head_dim)
        keys = keys.transpose(1, 2)
        queries = queries.transpose(1, 2)
        values = values.transpose(1, 2)

        # Nedensel maskeyle ölçeklenmiş nokta çarpımı dikkatini (öz-dikkat) hesapla
        attn_scores = queries @ keys.transpose(2, 3)  # Dot product for each head

        # Orijinal maske, token sayısına kırpılıp boole değere dönüştürüldü
        mask_bool = self.mask.bool()[:num_tokens, :num_tokens]

        # Dikkat skorlarını doldurmak için maskeyi kullan
        attn_scores.masked_fill_(mask_bool, -torch.inf)
        
        attn_weights = torch.softmax(attn_scores / keys.shape[-1]**0.5, dim=-1)
        attn_weights = self.dropout(attn_weights)

        # Boyut: (b, num_tokens, num_heads, head_dim)
        context_vec = (attn_weights @ values).transpose(1, 2) 
        
        # Başları birleştir; burada self.d_out = self.num_heads * self.head_dim
        context_vec = context_vec.contiguous().view(b, num_tokens, self.d_out)
        context_vec = self.out_proj(context_vec) # optional projection

        return context_vec

torch.manual_seed(123)

batch_size, context_length, d_in = batch.shape
d_out = 2
mha = MultiHeadAttention(d_in, d_out, context_length, 0.0, num_heads=2)

context_vecs = mha(batch)

print(context_vecs)
print("context_vecs.shape:", context_vecs.shape)
```

- Yukarıdakinin esasen `MultiHeadAttentionWrapper` sınıfının daha verimli, yeniden yazılmış bir sürümü olduğunu unutmayın
- Rastgele ağırlık başlatmaları farklı olduğu için elde edilen çıktı biraz farklı görünür; ancak her ikisi de ilerideki bölümlerde uygulayacağımız GPT sınıfında kullanılabilecek, tamamen işlevsel uygulamalardır

---

**Çıktı boyutları hakkında bir not**

- Yukarıdaki `MultiHeadAttention` sınıfında, daha önceki `MultiHeadAttentionWrapper` sınıfıyla aynı ayarı kullanmak için `d_out=2` kullandım
- `MultiHeadAttentionWrapper`, birleştirme (concatenation) nedeniyle `d_out * num_heads` çıktı başı boyutunu döndürür (yani `2*2 = 4`)
- Ancak `MultiHeadAttention` sınıfı (daha kullanıcı dostu olması için) çıktı başı boyutunu doğrudan `d_out` üzerinden denetlememize izin verir; yani `d_out = 2` ayarlarsak, baş sayısından bağımsız olarak çıktı başı boyutu 2 olur
- Geriye dönüp bakıldığında, okurların da [belirttiği gibi](https://github.com/rasbt/LLMs-from-scratch/pull/859), `MultiHeadAttention` sınıfını `d_out = 4` ile kullanmak daha sezgisel olabilirdi; böylece `d_out = 2` ile kullanılan `MultiHeadAttentionWrapper` ile aynı çıktı boyutlarını üretirdi.

---

- Ayrıca yukarıdaki `MultiHeadAttention` sınıfına bir doğrusal izdüşüm katmanı (`self.out_proj `) eklediğimizi unutmayın. Bu, boyutları değiştirmeyen basit bir doğrusal dönüşümdür. LLM uygulamalarında böyle bir izdüşüm katmanı kullanmak yerleşik bir uygulamadır, ancak kesinlikle zorunlu değildir (yakın tarihli araştırmalar, modelleme başarımını etkilemeden kaldırılabileceğini göstermiştir; bu bölümün sonundaki ileri okuma kısmına bakın)

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch03_compressed/26.webp" width="400px">

- Yukarıdakinin derli toplu ve verimli bir uygulamasıyla ilgileniyorsanız, PyTorch'taki [`torch.nn.MultiheadAttention`](https://pytorch.org/docs/stable/generated/torch.nn.MultiheadAttention.html) sınıfını da değerlendirebilirsiniz

- Yukarıdaki uygulama ilk bakışta biraz karmaşık görünebileceği için, `attn_scores = queries @ keys.transpose(2, 3)` çalıştırıldığında neler olduğuna bakalım:

```python
# (b, num_heads, num_tokens, head_dim) = (1, 2, 3, 4)
a = torch.tensor([[[[0.2745, 0.6584, 0.2775, 0.8573],
                    [0.8993, 0.0390, 0.9268, 0.7388],
                    [0.7179, 0.7058, 0.9156, 0.4340]],

                   [[0.0772, 0.3565, 0.1479, 0.5331],
                    [0.4066, 0.2318, 0.4545, 0.9737],
                    [0.4606, 0.5159, 0.4220, 0.5786]]]])

print(a @ a.transpose(2, 3))
```

- Bu durumda PyTorch'taki matris çarpımı uygulaması 4 boyutlu girdi tensörünü, matris çarpımı son 2 boyut (num_tokens, head_dim) arasında yapılacak ve ardından tek tek başlar için tekrarlanacak şekilde ele alır

- Örneğin, aşağıdaki, her baş için matris çarpımını ayrı ayrı hesaplamanın daha derli toplu bir yolu hâline gelir:

```python
first_head = a[0, 0, :, :]
first_res = first_head @ first_head.T
print("First head:\n", first_res)

second_head = a[0, 1, :, :]
second_res = second_head @ second_head.T
print("\nSecond head:\n", second_res)
```
