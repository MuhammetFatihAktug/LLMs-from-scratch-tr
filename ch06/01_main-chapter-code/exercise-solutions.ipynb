## Alıştırma 6.2: Modelin tamamına ince ayar yapmak

Yalnızca son transformer bloğuna ince ayar yapmak yerine, aşağıdaki satırları koddan kaldırarak modelin tamamına ince ayar yapabiliriz:

```python
for param in model.parameters():
    param.requires_grad = False
```

Kolaylık olsun diye bu deneyi, [../02_bonus_additional-experiments](../02_bonus_additional-experiments) klasöründeki kodu kullanarak şu komutla çalıştırabilirsiniz:

```bash
python additional-experiments.py --trainable_layers all
```

Bu, %1 daha iyi bir test doğruluğu verir: %96,67 (ana bölümdeki %95,67'ye karşılık).
